```python
import pandas as pd
import requests

# Your exact city and state configuration
locations = [
    {"city": "Amaravati", "state": "Andhra Pradesh", "lat": 16.5062, "lon": 80.6480},
    {"city": "Visakhapatnam", "state": "Andhra Pradesh", "lat": 17.6868, "lon": 83.2185},
    {"city": "Guwahati", "state": "Assam", "lat": 26.1445, "lon": 91.7362},
    {"city": "Patna", "state": "Bihar", "lat": 25.5941, "lon": 85.1376},
    {"city": "Chandigarh", "state": "Chandigarh", "lat": 30.7333, "lon": 76.7794},
    {"city": "Delhi", "state": "Delhi", "lat": 28.6139, "lon": 77.2090},
    {"city": "Ahmedabad", "state": "Gujarat", "lat": 23.0225, "lon": 72.5714},
    {"city": "Gurugram", "state": "Haryana", "lat": 28.4595, "lon": 77.0266},
    {"city": "Jorapokhar", "state": "Jharkhand", "lat": 23.7380, "lon": 86.4170},
    {"city": "Bengaluru", "state": "Karnataka", "lat": 12.9716, "lon": 77.5946},
    {"city": "Ernakulam", "state": "Kerala", "lat": 9.9816, "lon": 76.2999},
    {"city": "Kochi", "state": "Kerala", "lat": 9.9312, "lon": 76.2673},
    {"city": "Thiruvananthapuram", "state": "Kerala", "lat": 8.5241, "lon": 76.9366},
    {"city": "Mumbai", "state": "Maharashtra", "lat": 19.0760, "lon": 72.8777},
    {"city": "Shillong", "state": "Meghalaya", "lat": 25.5788, "lon": 91.8933},
    {"city": "Bhopal", "state": "Madhya Pradesh", "lat": 23.2599, "lon": 77.4126},
    {"city": "Aizawl", "state": "Mizoram", "lat": 23.7271, "lon": 92.7176},
    {"city": "Brajrajnagar", "state": "Odisha", "lat": 21.8159, "lon": 83.9213},
    {"city": "Talcher", "state": "Odisha", "lat": 20.9538, "lon": 85.2285},
    {"city": "Amritsar", "state": "Punjab", "lat": 31.6340, "lon": 74.8723},
    {"city": "Jaipur", "state": "Rajasthan", "lat": 26.9124, "lon": 75.7873},
    {"city": "Hyderabad", "state": "Telangana", "lat": 17.3850, "lon": 78.4867},
    {"city": "Chennai", "state": "Tamil Nadu", "lat": 13.0827, "lon": 80.2707},
    {"city": "Coimbatore", "state": "Tamil Nadu", "lat": 11.0168, "lon": 76.9558},
    {"city": "Lucknow", "state": "Uttar Pradesh", "lat": 26.8467, "lon": 80.9462},
    {"city": "Kolkata", "state": "West Bengal", "lat": 22.5726, "lon": 88.3639}
]

all_data = []

# Fetching the historical time range (e.g., 2015-04-01 to 2017-11-25)
for item in locations:
    url = (
        f"https://archive-api.open-meteo.com/v1/archive?"
        f"latitude={item['lat']}&longitude={item['lon']}&"
        f"start_date=2015-04-01&end_date=2017-11-25&"
        f"hourly=temperature_2m,relative_humidity_2m,wind_speed_10m"
    )
    response = requests.get(url).json()

    if "hourly" in response:
        df_loc = pd.DataFrame({
            "City": item["city"],
            "State": item["state"],
            "Datetime": response["hourly"]["time"],
            "Temperature_C": response["hourly"]["temperature_2m"],
            "Relative_Humidity_pct": response["hourly"]["relative_humidity_2m"],
            "Wind_Speed_kmh": response["hourly"]["wind_speed_10m"]
        })
        all_data.append(df_loc)

# Compile into your final dataframe
final_dataset = pd.concat(all_data, ignore_index=True)
print(final_dataset.head())
```
